## 0.3 Lists, indexing and loops
A list holds several values. Python starts counting positions at **zero**. A loop repeats an action; indentation marks the lines inside it.

```python
sequences = ["ACGT", "GGCC", "ATAT"]
print("First sequence:", sequences[0])
for sequence in sequences:
    print(sequence, "has", len(sequence), "bases")
```

## 0.4 Functions package a reusable calculation
Here `def` defines a function. Its input is a sequence and its output is the fraction of G and C bases. Our examples contain only A, C, G and T.

```python
def gc_fraction(sequence):
    return (sequence.count("G") + sequence.count("C")) / len(sequence)

print(gc_fraction("GGCA"))
```

## 0.5 Tables and arrays
A table has named columns. An array contains numbers arranged along axes. Its **shape** tells us how many entries each axis has.

In our first model, a row will be one sample and a column will be one measured feature. Here the values and labels are invented only to explain the format.

```python
table = pd.DataFrame({"sample": ["A", "B", "C"],
                      "measurement_1": [1.2, 2.0, 0.8],
                      "measurement_2": [5.0, 3.1, 4.2],
                      "label": [0, 1, 0]})
display(table)
x = table[["measurement_1", "measurement_2"]].to_numpy()
print("Array shape:", x.shape)
```
